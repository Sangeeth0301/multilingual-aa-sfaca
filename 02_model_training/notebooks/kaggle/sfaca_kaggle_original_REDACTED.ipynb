```python
import os, sys, subprocess
import pandas as pd

print("==========================================")
print("🚀 SFA-CA MASTER RUN: TRAIN ALL 5 ADAPTERS & EVALUATE MATRIX")
print("==========================================\n")

# 1. Setup & Environment
os.chdir("/kaggle/working/multilingual-aa-sfaca")
!git pull origin main

sys.path.insert(0, "/kaggle/working/multilingual-aa-sfaca/src")
!pip install -q "torchao>=0.16.0"

DATA_PATH = "/kaggle/input/datasets/sandeep01321/dataset/multitude_v3_clean.csv"
clusters = ["latin", "cyrillic", "greek", "arabic", "hanzi"]

# Create output directories
for c in clusters:
    os.makedirs(f"results/adapters/{c}", exist_ok=True)
os.makedirs("results/eval", exist_ok=True)

# 2. Train All 5 Adapters Sequentially (~10 min total)
for cluster in clusters:
    print(f"\n------------------------------------------")
    print(f"🔥 TRAINING ADAPTER: {cluster.upper()}")
    print(f"------------------------------------------\n")
    
    cmd = f"python src/train.py --cluster {cluster} --data {DATA_PATH} --epochs 3 --batch_size 32 --max_length 128 --lr 2e-4"
    subprocess.run(cmd, shell=True, check=True)

print("\n==========================================")
print("✅ ALL 5 ADAPTERS TRAINED SUCCESSFULLY!")
print("==========================================\n")

# 3. Run 5x5 Cross-Cluster Evaluation Matrix (~2 min)
print("📊 STARTING 5x5 CROSS-CLUSTER EVALUATION MATRIX...\n")

for train_c in clusters:
    for test_c in clusters:
        print(f"Evaluating Adapter [{train_c}] on Test Data [{test_c}]...")
        cmd = f"python src/evaluate_cross_cluster.py --data {DATA_PATH} --train_cluster {train_c} --test_cluster {test_c} --adapter_dir results/adapters --batch_size 32 --max_length 128"
        subprocess.run(cmd, shell=True)

print("\n==========================================")
print("📊 GENERATING FINAL 5x5 RESULTS MATRIX TABLE")
print("==========================================\n")

eval_dir = "results/eval"
records = []
if os.path.exists(eval_dir):
    for file in os.listdir(eval_dir):
        if file.endswith(".csv"):
            df = pd.read_csv(os.path.join(eval_dir, file))
            records.append(df)

if records:
    full_df = pd.concat(records, ignore_index=True)
    pivot_f1 = full_df.pivot(index="train_cluster", columns="test_cluster", values="f1_macro")
    
    print("==========================================")
    print("🏆 FINAL CROSS-CLUSTER F1 MACRO MATRIX TABLE")
    print("==========================================")
    print(pivot_f1.to_string())
    print("==========================================")
    
    pivot_f1.to_csv("results/cross_cluster_f1_matrix.csv")
    print("\n✅ Saved 5x5 Matrix to results/cross_cluster_f1_matrix.csv")
```

```python
import os, sys, subprocess, pandas as pd

os.chdir("/kaggle/working/multilingual-aa-sfaca")
sys.path.insert(0, "src")

clusters = ["latin", "cyrillic", "greek", "arabic", "hanzi"]
AUG_DATA = "/kaggle/working/multitude_10class.csv"

# PEFT patch (safe import)
try:
    import peft.tuners.lora.torchao as _lt
    _lt.dispatch_torchao = lambda *a, **kw: None
except Exception:
    pass

matrix = {c: {} for c in clusters}

print("📊 Running 5x5 Cross-Cluster Evaluation Matrix...\n")

for train in clusters:
    for test in clusters:
        cmd = (
            f"python src/evaluate_cross_cluster.py "
            f"--data {AUG_DATA} --train_cluster {train} --test_cluster {test}"
        )
        proc = subprocess.run(cmd, shell=True, capture_output=True, text=True)
        
        # Extract F1 Macro score from output
        f1 = 0.0
        for line in proc.stdout.split("\n"):
            if "F1 Macro:" in line:
                try:
                    f1 = float(line.split(":")[1].strip())
                except:
                    pass
        
        matrix[train][test] = f1
        print(f"Train: {train.upper():<10} | Test: {test.upper():<10} | F1 Macro: {f1:.4f}")

# Convert to 5x5 DataFrame
matrix_df = pd.DataFrame(matrix).T
matrix_df.index.name = "Train / Test"

print("\n" + "="*60)
print("🏆 FINAL 5x5 SFA-CA CROSS-CLUSTER MACRO F1 MATRIX")
print("="*60)
print(matrix_df.to_string())
print("="*60)

# Save to CSV
os.makedirs("results", exist_ok=True)
matrix_df.to_csv("results/cross_cluster_matrix.csv")
print("\n✅ Saved matrix to results/cross_cluster_matrix.csv")
```
